```python
# Step 1: Load Dataset
import pandas as pd

file_path = '../data/telecom_data.csv'  # Replace this with your file path

try:
    df = pd.read_csv(file_path)
    print("Dataset loaded successfully!")
    print("\nFirst 5 rows of the dataset:")
    print(df.head())
except FileNotFoundError:
    print(f"Error: File not found at {file_path}")
except pd.errors.EmptyDataError:
    print("Error: The file is empty.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
import matplotlib.pyplot as plt

# Step 2: Identify Top 10 Handsets
top_10_handsets = df['Handset Type'].value_counts().head(10)

print("\nTop 10 Handsets Used by Customers:")
print(top_10_handsets)

# Add this line to render plots in the notebook
%matplotlib inline

# Plot the Top 10 Handsets
plt.figure(figsize=(10, 6))
top_10_handsets.plot(kind='bar', color='skyblue')
plt.title("Top 10 Handsets Used by Customers")
plt.xlabel("Handset Type")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Step 3: Top 3 Handset Manufacturers
top_3_manufacturers = df['Handset Manufacturer'].value_counts().head(3)

print("\nTop 3 Handset Manufacturers:")
print(top_3_manufacturers)

# Render Matplotlib plots in the notebook
%matplotlib inline

# Plot the Top 3 Handset Manufacturers
plt.figure(figsize=(8, 5))  # Adjust figure size
top_3_manufacturers.plot(kind='bar', color='lightgreen')
plt.title("Top 3 Handset Manufacturers")
plt.xlabel("Handset Manufacturer")
plt.ylabel("Count")
plt.xticks(rotation=45)  # Rotate labels for readability
plt.show()
```

```python
import matplotlib.pyplot as plt

# Step 4: Top 5 Handsets per Top 3 Manufacturers
print("\nTop 5 Handsets per Top 3 Manufacturers:")

# Render plots inline
%matplotlib inline

# Loop through each top manufacturer and plot the top 5 handsets
for manufacturer in top_3_manufacturers.index:
    # Get the top 5 handsets for the current manufacturer
    top_5_handsets = df[df['Handset Manufacturer'] == manufacturer]['Handset Type'].value_counts().head(5)
    
    # Print the results
    print(f"\nTop 5 Handsets for Manufacturer: {manufacturer}")
    print(top_5_handsets)
    
    # Plot the results as a bar chart
    plt.figure(figsize=(8, 5))
    top_5_handsets.plot(kind='bar', color='coral')
    plt.title(f"Top 5 Handsets for {manufacturer}")
    plt.xlabel("Handset Type")
    plt.ylabel("Count")
    plt.xticks(rotation=45)  # Rotate labels for better readability
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Step 5: Aggregate User Behavior Analysis

# Group by 'MSISDN/Number' to analyze user behavior
user_behavior = df.groupby('MSISDN/Number').agg({
    'Dur. (ms)': 'sum',                       # Total session duration
    'Total UL (Bytes)': 'sum',                # Total upload data
    'Total DL (Bytes)': 'sum',                # Total download data
    'Social Media DL (Bytes)': 'sum',         # Social Media data
    'Google DL (Bytes)': 'sum',               # Google data
    'Email DL (Bytes)': 'sum',                # Email data
    'Youtube DL (Bytes)': 'sum',              # YouTube data
    'Netflix DL (Bytes)': 'sum',              # Netflix data
    'Gaming DL (Bytes)': 'sum',               # Gaming data
    'Other DL (Bytes)': 'sum'                 # Other data
}).reset_index()

# Calculate total data volume (UL + DL)
user_behavior['Total Data Volume (Bytes)'] = user_behavior['Total UL (Bytes)'] + user_behavior['Total DL (Bytes)']

# Display the first few rows
print("\nAggregated User Behavior Data:")
print(user_behavior.head())

# Summary statistics for numeric columns
print("\nSummary Statistics for User Behavior Data:")
print(user_behavior.describe())

# Plot the Total Data Volume Distribution
plt.figure(figsize=(10, 6))
sns.histplot(user_behavior['Total Data Volume (Bytes)'], bins=30, kde=True, color='skyblue')
plt.title("Distribution of Total Data Volume per User")
plt.xlabel("Total Data Volume (Bytes)")
plt.ylabel("Number of Users")
plt.show()

# Plot Top 10 Users by Total Data Volume
top_10_users = user_behavior.nlargest(10, 'Total Data Volume (Bytes)')

plt.figure(figsize=(10, 6))
sns.barplot(x='MSISDN/Number', y='Total Data Volume (Bytes)', data=top_10_users, palette='viridis')
plt.title("Top 10 Users by Total Data Volume")
plt.xlabel("User MSISDN/Number")
plt.ylabel("Total Data Volume (Bytes)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Step 6.1: Missing Value Treatment

# Check for missing values before treatment
print("\nMissing Values Before Treatment:")
print(user_behavior.isnull().sum())

# Replace missing values with the column mean
user_behavior = user_behavior.fillna(user_behavior.mean())

# Verify that missing values are treated
print("\nMissing Values After Treatment:")
print(user_behavior.isnull().sum())

print("\nMissing values replaced with column means.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Step 6.2: Segment Users into Decile Classes Based on Total Duration

# Create decile classes using 'Dur. (ms)' to split users into 5 groups
user_behavior['Decile'] = pd.qcut(user_behavior['Dur. (ms)'], 5, labels=False)

# Compute the total data volume (DL + UL) per decile
decile_data = user_behavior.groupby('Decile')['Total Data Volume (Bytes)'].sum()

# Display decile data
print("\nTotal Data Volume per Decile Class:")
print(decile_data)

# Visualize the Total Data Volume per Decile
plt.figure(figsize=(10, 6))
sns.barplot(x=decile_data.index, y=decile_data.values, palette="Blues_d")
plt.title("Total Data Volume per Duration Decile")
plt.xlabel("Duration Decile")
plt.ylabel("Total Data Volume (Bytes)")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Step 6.3: Univariate Analysis - Histograms and Boxplots

# List of key numeric columns for univariate analysis
columns_to_analyze = ['Dur. (ms)', 'Total UL (Bytes)', 'Total DL (Bytes)', 'Total Data Volume (Bytes)']

# Loop through each column and generate histograms and boxplots
for col in columns_to_analyze:
    plt.figure(figsize=(12, 5))
    
    # Histogram with KDE
    plt.subplot(1, 2, 1)
    sns.histplot(user_behavior[col], bins=20, kde=True, color='skyblue')
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    
    # Boxplot to check for outliers
    plt.subplot(1, 2, 2)
    sns.boxplot(x=user_behavior[col], color='lightcoral')
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Step 6.4: Bivariate Analysis - Relationship Between Applications and Total Data Volume

# List of key applications to analyze against Total Data Volume
applications = [
    'Social Media DL (Bytes)', 'Google DL (Bytes)', 'Email DL (Bytes)',
    'Youtube DL (Bytes)', 'Netflix DL (Bytes)', 'Gaming DL (Bytes)', 'Other DL (Bytes)'
]

# Loop through each application to generate scatter plots
for app in applications:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=user_behavior[app], y=user_behavior['Total Data Volume (Bytes)'], alpha=0.6, color='teal')
    plt.title(f"{app} vs Total Data Volume")
    plt.xlabel(app)
    plt.ylabel("Total Data Volume (Bytes)")
    plt.show()
```

```python
# Step 6.5: Correlation Analysis
selected_columns = [
    'Social Media DL (Bytes)', 'Google DL (Bytes)', 'Email DL (Bytes)',
    'Youtube DL (Bytes)', 'Netflix DL (Bytes)', 'Gaming DL (Bytes)', 'Other DL (Bytes)'
]
correlation_matrix = user_behavior[selected_columns].corr()

# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix for Application Data")
plt.show()
```

```python
# Step 6.5: Verify Selected Columns Exist in the DataFrame Before Analysis

# List of selected columns for correlation analysis
selected_columns = [
    'Social Media DL (Bytes)', 'Google DL (Bytes)', 'Email DL (Bytes)',
    'Youtube DL (Bytes)', 'Netflix DL (Bytes)', 'Gaming DL (Bytes)', 'Other DL (Bytes)'
]

# Check if selected columns exist in the DataFrame
missing_columns = [col for col in selected_columns if col not in user_behavior.columns]

if missing_columns:
    print("\nWarning: The following columns are missing from the DataFrame:")
    for col in missing_columns:
        print(f" - {col}")
    print("\nPlease verify the column names or update the selected_columns list.")
else:
    print("All selected columns are available for correlation analysis.")

# Display available columns for verification
print("\nAvailable Columns in the DataFrame:")
print(user_behavior.columns.tolist())

```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Step 7: Principal Component Analysis (PCA)

# Verify if all selected columns exist
selected_columns = [
    'Social Media DL (Bytes)', 'Google DL (Bytes)', 'Email DL (Bytes)',
    'Youtube DL (Bytes)', 'Netflix DL (Bytes)', 'Gaming DL (Bytes)', 'Other DL (Bytes)'
]

# Check for missing columns
missing_columns = [col for col in selected_columns if col not in user_behavior.columns]
if missing_columns:
    print(f"\nWarning: The following columns are missing: {missing_columns}")
else:
    print("All selected columns are available for PCA analysis.\n")

# Standardize the data
print("Standardizing the data...")
scaler = StandardScaler()
scaled_data = scaler.fit_transform(user_behavior[selected_columns])

# Apply PCA
print("Applying PCA...")
pca = PCA(n_components=2)  # Reduce to 2 components
pca_result = pca.fit_transform(scaled_data)

# Print explained variance ratio
print("\nExplained Variance Ratio by Components:")
for i, variance in enumerate(pca.explained_variance_ratio_):
    print(f"Component {i+1}: {variance:.4f}")

# Plot PCA results
plt.figure(figsize=(10, 6))
plt.scatter(pca_result[:, 0], pca_result[:, 1], alpha=0.6, color='purple')
plt.title("PCA: Application Data (Reduced to 2 Dimensions)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.show()
```
